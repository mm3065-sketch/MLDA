```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LinearRegression
from sklearn.metrics import accuracy_score, mean_squared_error, r2_score

data = pd.read_csv("/content/Employee_Salary_Dataset.csv")
print(data.head())

# Label encode the target variable 'Salary'
le = LabelEncoder()
data["Salary"] = le.fit_transform(data["Salary"])
print(data.head())

# Define features (x) and target variable (y)
# Exclude 'ID' as it's not a predictive feature and 'Salary' as it's the target
x = data[["Experience_Years", "Age", "Gender"]]
y = data["Salary"]

# Apply one-hot encoding to the 'Gender' column
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), ['Gender'])], remainder='passthrough')
x = ct.fit_transform(x)

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=2)

lrr = LinearRegression()
lrr.fit(x_train, y_train)
y_lrr_pred = lrr.predict(x_test)

mse = mean_squared_error(y_test, y_lrr_pred)
r2 = r2_score(y_test, y_lrr_pred)

print("Linear Regression MSE:", mse)
print("Linear Regression R2:", r2)
```
